### 2.2 Conditional statements

```python
def grade(marks):
    if marks >= 90: return "A+"
    elif marks >= 75: return "A"
    elif marks >= 60: return "B"
    elif marks >= 40: return "C"
    return "Fail"

for m in (95, 78, 55, 30):
    print(m, "->", grade(m))
```

### 2.3 Loops

```python
total = 0
for i in range(1, 11):
    total += i
print("Sum 1..10 =", total)

n, steps = 27, 0                      # while loop – Collatz steps
while n != 1:
    n = n // 2 if n % 2 == 0 else 3 * n + 1
    steps += 1
print("Collatz steps for 27:", steps)

print([x**2 for x in range(1, 8) if x % 2 == 1])   # list comprehension
```

### 2.3 Functions

```python
def is_prime(n: int) -> bool:
    """Return True if n is a prime number."""
    if n < 2: return False
    return all(n % i for i in range(2, int(n**0.5) + 1))

def summarize(values, *, round_to=2):
    """Return (min, max, mean) of a list of numbers."""
    return min(values), max(values), round(sum(values)/len(values), round_to)

print([p for p in range(1, 40) if is_prime(p)])
print(summarize([12, 7, 19, 3, 25]))
```

## 3. Statistical Foundations

```python
import numpy as np, pandas as pd, statistics as st
import matplotlib.pyplot as plt, seaborn as sns
sns.set_theme(style="whitegrid")

data = [45, 52, 58, 61, 61, 67, 70, 72, 75, 78, 81, 85, 88, 92, 98]
print("Mean     :", round(st.mean(data), 2))
print("Median   :", st.median(data))
print("Mode     :", st.mode(data))
print("Variance :", round(st.variance(data), 2), "(sample)")
print("Std dev  :", round(st.stdev(data), 2))
print("Percentiles 25/50/75/90:", np.percentile(data, [25, 50, 75, 90]))
```

```python
# Mean, variance and std computed from scratch to show understanding
n = len(data)
mean = sum(data) / n
var = sum((x - mean) ** 2 for x in data) / (n - 1)
print(round(mean, 2), round(var, 2), round(var ** 0.5, 2))   # matches the library values above
```

**Probability basics** – simulation vs. theory.

```python
rng = np.random.default_rng(42)
rolls = rng.integers(1, 7, 100_000)
print("P(rolling a 6): theory =", round(1/6, 4), "| simulated =", round((rolls == 6).mean(), 4))

d1, d2 = rng.integers(1, 7, 100_000), rng.integers(1, 7, 100_000)
print("P(sum = 7): theory =", round(6/36, 4), "| simulated =", round(((d1 + d2) == 7).mean(), 4))

# conditional probability P(sum>=10 | first die = 6)
cond = ((d1 + d2) >= 10)[d1 == 6].mean()
print("P(sum>=10 | first=6): theory = 0.5 | simulated =", round(cond, 3))
```

**Correlation** – how two variables move together.

```python
hours = rng.uniform(1, 10, 80)
score = 35 + 6 * hours + rng.normal(0, 6, 80)
r = np.corrcoef(hours, score)[0, 1]
print("Pearson r (study hours vs score):", round(r, 3))

plt.figure(figsize=(6,4)); sns.regplot(x=hours, y=score)
plt.xlabel("Study hours"); plt.ylabel("Score"); plt.title(f"Correlation r = {r:.2f}")
plt.tight_layout(); plt.savefig("data/correlation.png", dpi=120); plt.show()
```

## 5. Data Science Best Practices
Followed throughout this notebook:
- **Clean code & naming:** descriptive names (`study_hours`, `is_prime`), small functions, docstrings, type hints.
- **Reproducibility:** fixed random seed (`default_rng(42)`), `requirements.txt`, notebook runs top-to-bottom.
- **Proper data handling:** validate before use, keep raw data separate from cleaned data, never overwrite originals.
- **Documentation:** markdown explanation before each section and comments for non-obvious steps.
- **Responsible interpretation:** correlation is not causation; report uncertainty; avoid over-claiming from small samples.

```python
def validate_marks(values):
    """Best practice: validate input before analysis."""
    bad = [v for v in values if not (0 <= v <= 100)]
    if bad: raise ValueError(f"Invalid marks found: {bad}")
    return True

print(validate_marks([55, 80, 99]))
try: validate_marks([55, 180])
except ValueError as e: print("Caught:", e)
```

## 6. Fundamentals Assignments

```python
rng = np.random.default_rng(7)
n = 60
students = pd.DataFrame({
    "student_id": range(1, n + 1),
    "department": rng.choice(["AI&DS", "CSE", "ECE", "EEE"], n),
    "gender": rng.choice(["F", "M"], n),
    "study_hours": rng.uniform(1, 9, n).round(1),
    "attendance": rng.integers(55, 100, n),
})
students["marks"] = (30 + 5.5 * students["study_hours"] + 0.25 * students["attendance"] + rng.normal(0, 6, n)).clip(0, 100).round(1)
students.loc[[3, 17, 41], "marks"] = np.nan                       # missing values
students.to_csv("data/students.csv", index=False)
students.head()
```

### Assignment 1 – Python: grade report with functions and dictionaries

```python
def grade(m): return "A" if m >= 80 else "B" if m >= 65 else "C" if m >= 50 else "D"
report = {}
for _, row in students.dropna().iterrows():
    report[row["student_id"]] = grade(row["marks"])
from collections import Counter
print(Counter(report.values()))
```

### Assignment 2 – Statistics: descriptive summary by department

```python
print(students["marks"].describe().round(2))
print(students.groupby("department")["marks"].agg(["count", "mean", "median", "std"]).round(2))
print("Mode of attendance:", students["attendance"].mode().tolist())
```

### Assignment 3 – Data types: inspection and conversion

```python
students["department"] = students["department"].astype("category")
students["gender"] = students["gender"].astype("category")
print(students.dtypes); print("Memory:", students.memory_usage(deep=True).sum(), "bytes")
```

### Assignment 4 – Data manipulation: handle missing data, filter, sort

```python
print("Missing:", students["marks"].isna().sum())
students["marks"] = students["marks"].fillna(students["marks"].median())
top = students[(students["attendance"] >= 85) & (students["marks"] >= 75)].sort_values("marks", ascending=False)
print("High attendance & high marks:", len(top))
top.head()
```

### Assignment 5 – Basic analysis: relationships and visuals

```python
print(students[["study_hours", "attendance", "marks"]].corr().round(2))

fig, ax = plt.subplots(1, 3, figsize=(15, 4))
sns.histplot(students["marks"], bins=12, kde=True, ax=ax[0]).set(title="Marks distribution")
sns.boxplot(data=students, x="department", y="marks", ax=ax[1]).set(title="Marks by department")
sns.scatterplot(data=students, x="study_hours", y="marks", hue="department", ax=ax[2]).set(title="Study hours vs marks")
plt.tight_layout(); plt.savefig("data/assignment_plots.png", dpi=120); plt.show()
```

```python
r = students["study_hours"].corr(students["marks"])
best_dept = students.groupby("department", observed=True)["marks"].mean().idxmax()
print(f"Study hours and marks are strongly positively correlated (r = {r:.2f}).")
print(f"{best_dept} has the highest average marks in this sample.")
print("Caution: correlation does not prove causation, and n = 60 is a small sample.")
```
